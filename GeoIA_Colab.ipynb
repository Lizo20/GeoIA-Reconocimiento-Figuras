## 3. Ver resultados y gráficos

```python
import json
from IPython.display import Image, display
with open('resultados/metricas.json',encoding='utf-8') as f:
    metricas=json.load(f)
print('Exactitud en prueba sintética:',f"{metricas['exactitud_prueba']:.2%}")
print('Aciertos:',metricas['aciertos'],'/',metricas['muestras_prueba'])
display(Image('resultados/muestras.png'))
display(Image('resultados/curva_validacion.png'))
display(Image('resultados/matriz_confusion.png'))
```
